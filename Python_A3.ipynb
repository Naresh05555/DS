#Prime Number Program

```python
# Function to check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

# Test the function
number = 29
if is_prime(number):
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")

#Explanation:
#1.Function Definition: The is_prime function takes a number n as input
#2.Check for Numbers Less than or Equal to 1: Prime numbers are greater than 1, so numbers less than or equal to 1 are not prime.
#3.Loop through Possible Divisors: Loop through numbers from 2 to the square root of n If n is divisible by any of these, it's not prime.
#4.Return Result: If no divisors are found, the number is prime.
```

#Random Number Program


```python
import random
import numpy as np

# Function to generate random numbers and calculate their product
def product_of_random_numbers(count, lower_bound, upper_bound):
    # Generate a list of random numbers
    random_numbers = [random.randint(lower_bound, upper_bound) for _ in range(count)]
    # Calculate the product using numpy
    product = np.prod(random_numbers)
    return random_numbers, product

# Number of random numbers to generate
count = 5
lower_bound = 1
upper_bound = 10

# Get the random numbers and their product
random_numbers, product = product_of_random_numbers(count, lower_bound, upper_bound)

print("Random Numbers:", random_numbers)
print("Product:", product)

#Explanation:
#1.Import Libraries:
#random for generating random numbers.
#numpy for calculating the product of numbers.

#2.Function Definition:
#product_of_random_numbers(count, lower_bound, upper_bound) generates count random numbers between lower_bound and upper_bound, and calculates their product.

#3.Generate Random Numbers:
#Use a list comprehension to create a list of random integers within the specified bounds.

#4.Calculate the Product:
#Use numpy.prod to calculate the product of the list of random numbers.

#5.Example Usage:
#Specify the number of random numbers (count), and the range for the random numbers (lower_bound and upper_bound).
#Call the function to get the list of random numbers and their product.
#Print the random numbers and their product.
```

#Square Of Even/odd Number Program

```python
# Function to check if a number is even or odd and calculate its square
def square_of_number(number):
    if number % 2 == 0:
        result = f"The number {number} is even."
    else:
        result = f"The number {number} is odd."
    square = number ** 2
    return result, square

# Test the function
number = 7
result, square = square_of_number(number)
print(result)
print(f"The square of {number} is {square}.")

#Explanation:
#1.Function Definition:square_of_number(number) checks if the number is even or odd and calculates its square.
#2.Check Even or Odd:Use the modulus operator (%) to determine if the number is even (number % 2 == 0) or odd (number % 2 != 0).
#3.Calculate Square:The square of the number is calculated using the exponentiation operator (**).
#4.Return Result:The function returns whether the number is even or odd and its square.
#5.Example Usage:Replace number = 7 with any number you want to test. The program will output whether the number is even or odd and its square.
```

#write a program to count the number of words

```python
from collections import Counter

# Function to count the frequency of each word in a given text
def word_frequency(input_text):
    # Split the text into words
    words = input_text.split()
    # Count the frequency of each word
    word_count = Counter(words)
    return word_count

# Example usage
input_text = "This is a sample text. This text will be used to demonstrate the word counter."
word_count = word_frequency(input_text)

# Print the word frequencies
for word, count in word_count.items():
    print(f"'{word}': {count}")
#Explanation:
#1.Import Libraries:collections.Counter: For counting the frequency of each word.
#2.Function Definition:word_frequency(input_text) takes a string of text as input.
#3.Split Text into Words:Use the split() method to break the text into individual words.
#4.Count Word Frequency:Use Counter from the collections module to count the frequency of each word.
#5.Example Usage:The given input_text is passed to the word_frequency function, and the word frequencies are printed.
```

#Program For Palindrom

```python
# Function to check if a string is a palindrome
def is_palindrome(text):
    # Remove any spaces and convert the text to lowercase for uniformity
    cleaned_text = ''.join(text.split()).lower()
    # Check if the cleaned text reads the same forwards and backwards
    return cleaned_text == cleaned_text[::-1]

# Test the function
input_text = "A man a plan a canal Panama"
if is_palindrome(input_text):
    print(f'"{input_text}" is a palindrome.')
else:
    print(f'"{input_text}" is not a palindrome.')
#Explanation:
#1.Function Definition:is_palindrome(text) takes a string as input.
#2.Clean the Text:Remove spaces and convert the text to lowercase to ensure uniformity and disregard case sensitivity or spaces.
#3.Check Palindrome:Compare the cleaned text with its reverse (cleaned_text[::-1]).
#4.Example Usage:The input text "A man a plan a canal Panama" is tested for palindrome. The program outputs whether it is a palindrome.
```
