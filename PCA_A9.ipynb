#2.Dimensionality Reduction with PCA

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Standardize the features
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df.iloc[:, 1:])  # Skip the 'Type' column for PCA

# Step 2: Apply PCA
pca = PCA()
pca.fit(df_scaled)

# Step 3: Determine optimal number of components
explained_variance_ratio = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance_ratio)

# Scree plot
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio, marker='o', label='Individual Variance')
plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker='x', label='Cumulative Variance')
plt.axhline(y=0.95, color='r', linestyle='--', label='95% Threshold')
plt.title('Explained Variance by Principal Components')
plt.xlabel('Principal Component')
plt.ylabel('Variance Explained')
plt.legend()
plt.grid(True)
plt.show()

# Step 4: Transform the dataset
optimal_components = np.argmax(cumulative_variance >= 0.95) + 1  # Select components explaining 95% variance
pca = PCA(n_components=optimal_components)
data_pca = pca.fit_transform(df_scaled)

# Display results
print(f"Optimal Number of Components: {optimal_components}")
print(f"Transformed Dataset Shape: {data_pca.shape}")

#Explanation:
#1.Standardization:
#Features are scaled so that PCA focuses on variance rather than magnitude differences.
#StandardScaler ensures a mean of 0 and standard deviation of 1.

#2.Applying PCA:
#PCA is fit to the standardized dataset to compute the principal components.
#The explained_variance_ratio_ gives the proportion of variance explained by each component.

#3.Selecting Components:
#The scree plot visualizes how much variance each component explains.
#Cumulative variance helps to decide the number of components required to retain a significant portion (e.g., 95%) of the variance.

#4.Transforming Dataset:
#The dataset is transformed into a new space defined by the selected principal components.
```

#3.Clustering with Original Data:


```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Standardize the features
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df.iloc[:, 1:])  # Skip 'Type' for clustering

# Step 2: Apply K-means clustering for a range of cluster numbers
inertia = []
silhouette_scores = []

for k in range(2, 11):  # Range of clusters to test
    kmeans = KMeans(n_clusters=k, random_state=42)
    cluster_labels = kmeans.fit_predict(df_scaled)
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(df_scaled, cluster_labels))

# Plot Elbow Method
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia, marker='o', label='Inertia (Elbow)')
plt.title('Elbow Method for Optimal Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.legend()
plt.grid(True)
plt.show()

# Plot Silhouette Scores
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_scores, marker='x', label='Silhouette Score')
plt.title('Silhouette Score for Optimal Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.legend()
plt.grid(True)
plt.show()

# Step 3: Fit K-means with optimal clusters (e.g., k=3)
optimal_k = silhouette_scores.index(max(silhouette_scores)) + 2
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
cluster_labels = kmeans.fit_predict(df_scaled)

# Step 4: Visualize Clustering Results
df['Cluster'] = cluster_labels
sns.pairplot(df, hue='Cluster', palette='tab10', diag_kind='kde')
plt.suptitle("Cluster Visualization", y=1.02)
plt.show()

# Step 5: Evaluate Clustering Performance
db_index = davies_bouldin_score(df_scaled, cluster_labels)
silhouette_avg = silhouette_score(df_scaled, cluster_labels)

print(f"Optimal Number of Clusters: {optimal_k}")
print(f"Davies–Bouldin Index: {db_index:.2f}")
print(f"Silhouette Score: {silhouette_avg:.2f}")

#Explanation:
#1.Standardization:
#Necessary to avoid bias due to varying feature magnitudes.

#2.Determine Optimal Clusters:
#Inertia (Elbow Method): Measures how tight the clusters are. The optimal cluster count is often at the "elbow."
#Silhouette Score: Measures how well-separated the clusters are, with higher scores indicating better clustering.

#3.Fit K-Means:
#Once the optimal number of clusters is determined, fit the K-means model to create the clusters.

#4.Visualization:
#Use pair plots to view clusters in 2D projections of the feature space.


#5.Evaluation:
#Silhouette Score: Higher values are better.
#Davies–Bouldin Index: Lower values indicate better-defined clusters.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris

# Load Iris dataset
data = load_iris()
X = data.data
y = data.target

# Perform PCA (reduce to 2 components)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Apply KMeans clustering
kmeans = KMeans(n_clusters=3, random_state=42)
y_kmeans = kmeans.fit_predict(X_pca)

# Plot the results of PCA and KMeans
plt.figure(figsize=(10, 6))

# Scatter plot of PCA components with KMeans clusters
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_kmeans, cmap='viridis', marker='o', edgecolor='k', s=100, alpha=0.6)

# Plot the centroids of clusters
centroids = kmeans.cluster_centers_
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, marker='x', label='Centroids')

# Add labels and title
plt.title('PCA of Iris Dataset with KMeans Clustering')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()

# Show the plot
plt.show()

#Explanation:
#1.PCA:
#Reduces the dimensionality of the Iris dataset from 4 features to 2 components.
#PCA(n_components=2) keeps the top two principal components to visualize the data in 2D.

#2.K-Means Clustering:
#We apply the K-Means algorithm to the 2D PCA-transformed data, specifying n_clusters=3 (since there are 3 species in the Iris dataset).
#The clustering labels are used to color the points.

#3.Plotting:
#The scatter plot visualizes the data points in the reduced 2D space, with colors representing the clusters.
#The red x marks represent the centroids of the clusters.
```
