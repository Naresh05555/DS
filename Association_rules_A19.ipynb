```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Simulated Online Retail dataset
data = {
    "InvoiceNo": ["10001", "10001", "10002", "10002", "10003", "10004", "10004", "10005", "C10006"],
    "Description": ["Item A", "Item B", "Item A", "Item C", "Item D", "Item A", "Item B", "Item E", "Item F"],
    "Quantity": [2, 3, 1, 2, 5, 2, -1, 4, 1],
    "CustomerID": [101, 101, 102, 102, 103, 104, 104, 105, 106],
}

# Convert the dictionary to a DataFrame
df = pd.DataFrame(data)

# Step 1: Remove duplicates
df = df.drop_duplicates()

# Step 2: Filter transactions
# Remove canceled transactions (InvoiceNo starting with 'C')
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]

# Keep only positive quantities
df = df[df['Quantity'] > 0]

# Step 3: Prepare data for association rules
# Group by InvoiceNo and Description, summing the quantities
basket = df.groupby(['InvoiceNo', 'Description'])['Quantity'].sum().unstack().fillna(0)

# Convert quantities to binary (1 if purchased, 0 otherwise)
basket = basket.applymap(lambda x: 1 if x > 0 else 0)

# Plot 1: Bar plot for item frequency
item_frequencies = basket.sum(axis=0).sort_values(ascending=False)
plt.figure(figsize=(8, 6))
item_frequencies.plot(kind='bar', color='skyblue')
plt.title("Item Frequency", fontsize=16)
plt.xlabel("Items", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Plot 2: Heatmap for item correlations
plt.figure(figsize=(8, 6))
sns.heatmap(basket.corr(), annot=True, cmap="YlGnBu", fmt=".2f", linewidths=0.5)
plt.title("Correlation Between Items", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Sample data
import warnings
warnings.filterwarnings('ignore')
data = {'Transaction_ID': [1, 1, 2, 2, 3, 3, 4, 4, 4],
        'Product': ['Milk', 'Bread', 'Milk', 'Butter', 'Bread', 'Butter', 'Milk', 'Bread', 'Butter']}
df = pd.DataFrame(data)

# Convert the data into a binary matrix format
basket = df.pivot_table(index='Transaction_ID', columns='Product', aggfunc='size', fill_value=0)
print(basket)
```

```python
rules = association_rules(frequent_itemsets,metric="confidence", min_threshold=0.7, support_only=False)
```

```python
# Filter rules based on certain thresholds
filtered_rules = rules[(rules['support'] >= 0.3) &
                       (rules['confidence'] >= 0.7) &
                       (rules['lift'] >= 1.5)]

print(filtered_rules)

#Explanation:
#1.Data Preparation:
#The data consists of transactions where each transaction is represented by a product that a customer purchased. We need to convert this data into a binary matrix format for the Apriori algorithm to work.
#Input Data: The dataset contains two columns:
#Transaction_ID: The ID of each transaction (which represents a customer’s purchase).


#2. Apriori Algorithm:
#The Apriori algorithm is used to identify frequent itemsets (sets of products that are frequently bought together). We use the apriori() function from mlxtend to apply the algorithm to the binary matrix and extract frequent itemsets.
#min_support: This is the threshold for the support of itemsets. It specifies the minimum proportion of transactions in which an itemset must appear to be considered frequent. In this case, min_support=0.3 means that we only keep itemsets that appear in at least 30% of transactions.

#3.Association Rules:
#Once frequent itemsets are identified, the next step is to generate association rules. Association rules express relationships between itemsets
#Antecedent (left-hand side) is the set of items that imply the presence of another set.
#Consequent (right-hand side) is the set of items that are likely to be bought if the antecedent is purchased.
#metric:
#We use lift as the metric for evaluating the strength of the rules. Lift measures how much more likely the consequent is to occur when the antecedent occurs, compared to when it doesn't occur. A lift value greater than 1 indicates a positive association between the antecedent and consequent.
#min_threshold:
#This sets the minimum value of lift for a rule to be considered significant. In this case, min_threshold=1.0 means that we only keep rules where the lift is greater than or equal to 1.

#4.Output of Association Rules:
#The association_rules() function generates rules with columns for:
#antecedents: Items in the left-hand side of the rule (items that were bought first).
#consequents: Items in the right-hand side of the rule (items that are likely to be bought if the antecedents are bought).
#support: The proportion of transactions containing both antecedents and consequents.
#confidence: The proportion of transactions with the antecedents that also contain the consequents.
#lift: The strength of the association between the antecedents and consequents.

#5.Interpretation of Rules:
#Lift > 1: For example, the rule (Bread) => (Milk) with a lift of 2.22 means that when Bread is purchased, Milk is 2.22 times more likely to be bought than if Bread and Milk were independent.
#Confidence: In the rule (Bread) => (Milk), a confidence of 0.67 means that 67% of the transactions containing Bread also contain Milk.
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
import matplotlib.pyplot as plt

# Sample data
data = {'Transaction_ID': [1, 1, 2, 2, 3, 3, 4, 4, 4],
        'Product': ['Milk', 'Bread', 'Milk', 'Butter', 'Bread', 'Butter', 'Milk', 'Bread', 'Butter']}
df = pd.DataFrame(data)

# Convert the data into a binary matrix format
basket = df.pivot_table(index='Transaction_ID', columns='Product', aggfunc='size', fill_value=0)
print("Basket (Binary Matrix):")
print(basket)

# Apply Apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(basket, min_support=0.3, use_colnames=True)
print("\nFrequent Itemsets:")
print(frequent_itemsets)

# Generate association rules with lift as the metric
rules = association_rules(frequent_itemsets, num_itemsets=None, metric="confidence", min_threshold=0.7, support_only=False)
print("\nAssociation Rules:")
print(rules)

# Plotting the Association Rules
plt.figure(figsize=(10, 6))

# Scatter plot: X-axis = support, Y-axis = confidence, Color = lift
plt.scatter(rules['support'], rules['confidence'], c=rules['lift'], cmap='coolwarm', s=100, alpha=0.6)

# Adding titles and labels
plt.title('Association Rules: Support vs Confidence (Colored by Lift)', fontsize=14)
plt.xlabel('Support', fontsize=12)
plt.ylabel('Confidence', fontsize=12)

# Adding color bar to represent Lift
plt.colorbar(label='Lift')

# Display the plot
plt.show()
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
import matplotlib.pyplot as plt

# Sample data (transactions)
data = {'Transaction_ID': [1, 1, 2, 2, 3, 3, 4, 4, 4],
        'Product': ['Milk', 'Bread', 'Milk', 'Butter', 'Bread', 'Butter', 'Milk', 'Bread', 'Butter']}
df = pd.DataFrame(data)

# Convert the data into a binary matrix format (basket format)
basket = df.pivot_table(index='Transaction_ID', columns='Product', aggfunc='size', fill_value=0)
print("Basket (Binary Matrix):")
print(basket)

# Apply Apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(basket, min_support=0.3, use_colnames=True)
print("\nFrequent Itemsets:")
print(frequent_itemsets)

# Generate association rules with lift as the metric
rules = association_rules(frequent_itemsets, num_itemsets=None, metric="confidence", min_threshold=0.7, support_only=False)
print("\nAssociation Rules:")
print(rules)

# Analysis and Interpretation of Rules
print("\nAnalysis and Interpretation:")
for index, row in rules.iterrows():
    print(f"\nRule {index + 1}: If {row['antecedents']} => {row['consequents']}")
    print(f"  - Support: {row['support']:.2f}")
    print(f"  - Confidence: {row['confidence']:.2f}")
    print(f"  - Lift: {row['lift']:.2f}")
    if row['lift'] > 1:
        print("  - Interpretation: The items are strongly associated and bought together more often than by chance.")
    else:
        print("  - Interpretation: The items are not strongly associated.")

# Plotting the Association Rules
plt.figure(figsize=(10, 6))

# Scatter plot: X-axis = support, Y-axis = confidence, Color = lift
plt.scatter(rules['support'], rules['confidence'], c=rules['lift'], cmap='coolwarm', s=100, alpha=0.6)

# Adding titles and labels
plt.title('Association Rules: Support vs Confidence (Colored by Lift)', fontsize=14)
plt.xlabel('Support', fontsize=12)
plt.ylabel('Confidence', fontsize=12)

# Adding color bar to represent Lift
plt.colorbar(label='Lift')

# Display the plot
plt.show()

# Further Insights and Strategy Recommendations
print("\nFurther Insights and Strategy Recommendations:")
# Example insights based on the analysis
for index, row in rules.iterrows():
    if row['lift'] > 1:
        print(f"Rule {index + 1} suggests a strong association between {row['antecedents']} and {row['consequents']}.")
        if 'Bread' in str(row['antecedents']) and 'Milk' in str(row['consequents']):
            print("  - Insight: Bread and Milk are commonly bought together. Consider bundling them in promotions.")
        if 'Milk' in str(row['antecedents']) and 'Butter' in str(row['consequents']):
            print("  - Insight: Milk and Butter are commonly bought together. Target them with special offers.")
```

#Interview Questions
#1.What is lift and why is it important in Association rules?
#Ans:Lift is a key metric used in association rule learning to measure the strength of the relationship between two items. It is defined as the ratio of the observed support of an itemset to the expected support if the items were independent.

#2.What is support and Confidence. How do you calculate them?
#Ans:Support:
#Support measures how frequently an itemset appears in the dataset. It is the proportion of transactions in which an itemset appears.
#Calculation: If you have a dataset with 𝑁 transactions and an itemset 𝑋 appears in 𝑇 of these transactions, then the support of itemset 𝑋 is calculated as:
#Support(𝑋)=𝑇/𝑁

#Confidence:
#Confidence measures the likelihood that item 𝐵is also present in transactions that contain item 𝐴. It's the conditional probability of finding 𝐵 given that 𝐴is already present.
#Confidence(𝐴→𝐵)=Support(𝐴∩𝐵)Support(A)=𝑇/𝑆

#3.What are some limitations or challenges of Association rules mining?
#Ans:Association rule mining, while powerful, comes with its own set of limitations and challenges

#1.Scalability:
#Large datasets can lead to a combinatorial explosion in the number of potential itemsets, making the process computationally intensive and time-consuming.
#2.Interpretability:
#Generating a large number of rules can make it difficult to interpret and understand the significant ones. Filtering out irrelevant or less significant rules can be a challenge.
#3.Sparsity of Data:
#In datasets with a vast number of items but a relatively small number of transactions, many itemsets may have low support, making it harder to find meaningful associations.
#4.Choosing the Right Thresholds:
#Setting appropriate minimum support and confidence thresholds can be tricky. Too high a threshold might miss interesting rules, while too low a threshold may produce too many insignificant rules.
#5.Rule Redundancy:
#Many discovered rules can be redundant, as they might be derivable from other rules. This requires additional filtering and pruning.
#6.Binary Attributes Limitation:
#Traditional association rule mining often assumes binary attributes (presence or absence of an item). This can limit its application to datasets with numerical or categorical attributes.
#7.Context Ignorance:
#Association rules do not consider the context in which transactions occur. For example, they don’t account for time, location, or other relevant factors that might influence the associations.
#8.Actionability:
#Not all discovered associations are actionable or meaningful from a business perspective. The challenge lies in identifying the rules that can lead to practical insights and strategies.
