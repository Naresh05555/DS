```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.cluster.hierarchy import fcluster

# Step 1: Load the dataset
df = pd.read_excel(r"C:\Users\nares.LAP\OneDrive\Documents\excelr\Copy of EastWestAirlines.xlsx")

# Step 2: Check the columns and data types
print("Dataset columns and data types:\n", df.dtypes)

# Step 3: Preprocess the data
# Drop non-numeric columns and rows with missing values
df_clean = df.select_dtypes(include=[np.number]).dropna()

# Check if we have any numeric data after cleaning
if df_clean.empty:
    print("Dataframe has no numeric columns after cleaning!")
else:
    print("Dataframe after cleaning (first few rows):\n", df_clean.head())

# Step 4: If we still have valid data, continue
if not df_clean.empty:
    # Step 4: Normalize the data
    scaler = StandardScaler()
    df_scaled = scaler.fit_transform(df_clean)

    # Step 5: Apply PCA for dimensionality reduction (optional, for visualization)
    pca = PCA(n_components=2)
    df_pca = pca.fit_transform(df_scaled)

    # Step 6: K-Means Clustering
    # Elbow Method to find the optimal number of clusters (K)
    inertia = []
    for k in range(1, 11):
        kmeans = KMeans(n_clusters=k)
        kmeans.fit(df_scaled)
        inertia.append(kmeans.inertia_)

    # Plot the elbow curve
    plt.figure(figsize=(8, 6))
    plt.plot(range(1, 11), inertia, marker='o')
    plt.title('Elbow Method for Optimal K')
    plt.xlabel('Number of Clusters')
    plt.ylabel('Inertia')
    plt.show()

    # Based on the plot, choose K=3 (for example)
    kmeans = KMeans(n_clusters=3)
    kmeans_labels = kmeans.fit_predict(df_scaled)

    # Evaluate the clustering (Silhouette Score)
    sil_score_kmeans = silhouette_score(df_scaled, kmeans_labels)
    print(f"Silhouette Score for K-Means: {sil_score_kmeans}")

    # Step 7: Hierarchical Clustering
    Z = linkage(df_scaled, method='ward')  # 'ward' minimizes variance within clusters

    # Plot the dendrogram
    plt.figure(figsize=(10, 7))
    dendrogram(Z)
    plt.title('Hierarchical Clustering Dendrogram')
    plt.xlabel('Index')
    plt.ylabel('Distance')
    plt.show()

    # Apply hierarchical clustering and set max clusters to 3
    hierarchical_labels = fcluster(Z, t=3, criterion='maxclust')

    # Evaluate the clustering (Silhouette Score)
    sil_score_hierarchical = silhouette_score(df_scaled, hierarchical_labels)
    print(f"Silhouette Score for Hierarchical Clustering: {sil_score_hierarchical}")

    # Step 8: DBSCAN Clustering
    dbscan = DBSCAN(eps=0.5, min_samples=5)  # Experiment with different eps and min_samples
    dbscan_labels = dbscan.fit_predict(df_scaled)

    # DBSCAN may classify some points as noise (-1)
    noise_points = np.sum(dbscan_labels == -1)
    print(f"Number of noise points: {noise_points}")

    # Evaluate the clustering (Silhouette Score)
    sil_score_dbscan = silhouette_score(df_scaled, dbscan_labels)
    print(f"Silhouette Score for DBSCAN: {sil_score_dbscan}")

    # Step 9: Visualize the clusters

    # Visualize K-Means Clustering
    plt.figure(figsize=(8, 6))
    plt.scatter(df_pca[:, 0], df_pca[:, 1], c=kmeans_labels, cmap='viridis')
    plt.title('K-Means Clustering')
    plt.show()

    # Visualize Hierarchical Clustering
    plt.figure(figsize=(8, 6))
    plt.scatter(df_pca[:, 0], df_pca[:, 1], c=hierarchical_labels, cmap='viridis')
    plt.title('Hierarchical Clustering')
    plt.show()

    # Visualize DBSCAN Clustering
    plt.figure(figsize=(8, 6))
    plt.scatter(df_pca[:, 0], df_pca[:, 1], c=dbscan_labels, cmap='viridis')
    plt.title('DBSCAN Clustering')
    plt.show()

else:
    print("No numeric columns found. Please check the dataset for numeric data.")

#Explanation:
#1.K-Means Clustering:
#Objective: K-Means divides the dataset into K clusters based on the similarity of data points.
#Steps:
#Preprocess the data: Normalize or scale the data so that all features are on the same scale.
#Choose K: Use the Elbow Method to find the best number of clusters (K). You plot the "inertia" (the sum of squared distances between points and their cluster centers) for different K values and look for a point where the decrease in inertia slows down.
#Fit K-Means: After finding the optimal K, run the K-Means algorithm to create the clusters.
#Evaluate: Check the quality of clustering using the Silhouette Score (a measure of how similar a point is to its own cluster compared to other clusters).
#Experimentation: Try different values of K to see which one works best.

#2. Hierarchical Clustering
#Objective: Builds a tree of clusters (called a dendrogram) and doesn't require specifying the number of clusters upfront.
#Steps:
#Preprocess the data: Scale the data if necessary.
#Create the Dendrogram: Use the hierarchical clustering method (agglomerative is most common) and create a dendrogram. It shows the distances between clusters as they are merged.
#Choose clusters: Cut the dendrogram at a certain level to decide the number of clusters.
#Evaluate: Like K-Means, you can also evaluate the clusters with the Silhouette Score.
#Experimentation: Experiment with different linkage criteria for hierarchical clustering:
#Ward’s Method: Minimizes the variance within clusters.
#Single Linkage: Minimum distance between points in different clusters.
#Complete Linkage: Maximum distance between points in different clusters.

#3.DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
#Objective: DBSCAN forms clusters based on the density of data points. It also identifies outliers (points that don’t belong to any cluster).
#Steps:
#Preprocess the data: Scale the data if needed.
#Set parameters:
#epsilon (eps): The maximum distance between two points to be considered in the same cluster.
#minPts: The minimum number of points required to form a dense region (cluster).
#Run DBSCAN: Apply the DBSCAN algorithm to the data. It will assign points to clusters or label them as noise (-1).
#Experimentation: Experiment with different values of epsilon and minPts to see how they affect the clustering.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.datasets import make_blobs

# Step 1: Generate synthetic data (instead of loading from a file)
n_samples = 300
n_features = 5
n_clusters = 3

data, _ = make_blobs(n_samples=n_samples, n_features=n_features, centers=n_clusters, random_state=42)
df = pd.DataFrame(data, columns=[f'Feature_{i}' for i in range(n_features)])

# Step 2: Visualize data distribution using box plot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df)
plt.title('Box Plot of Numerical Features')
plt.xticks(rotation=90)
plt.show()

# Step 3: Normalize the data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# Step 4: Apply PCA for visualization (reduce to 2D for plotting)
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)

# Step 5: K-Means Clustering
# Find the optimal K using the Elbow Method
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df_scaled)
    inertia.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertia, marker='o')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.show()

# Fit K-Means with optimal K (e.g., K=3)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(df_scaled)
silhouette_kmeans = silhouette_score(df_scaled, kmeans_labels)
print(f'Silhouette Score for K-Means: {silhouette_kmeans:.2f}')

# Visualize K-Means clustering results
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=kmeans_labels, cmap='viridis')
plt.title('K-Means Clustering')
plt.show()

# Step 6: Hierarchical Clustering
Z = linkage(df_scaled, method='ward')

# Plot the dendrogram
plt.figure(figsize=(10, 6))
dendrogram(Z)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.show()

# Cut the dendrogram to form clusters
hierarchical_labels = fcluster(Z, t=3, criterion='maxclust')
silhouette_hierarchical = silhouette_score(df_scaled, hierarchical_labels)
print(f'Silhouette Score for Hierarchical Clustering: {silhouette_hierarchical:.2f}')

# Visualize Hierarchical clustering results
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=hierarchical_labels, cmap='viridis')
plt.title('Hierarchical Clustering')
plt.show()

# Step 7: DBSCAN Clustering
dbscan = DBSCAN(eps=0.8, min_samples=5)
dbscan_labels = dbscan.fit_predict(df_scaled)

# Evaluate DBSCAN (check for noise points)
num_noise_points = np.sum(dbscan_labels == -1)
print(f'Number of noise points in DBSCAN: {num_noise_points}')

# Only calculate silhouette score if there are more than 1 cluster (excluding noise)
if len(set(dbscan_labels)) > 1:
    silhouette_dbscan = silhouette_score(df_scaled, dbscan_labels)
    print(f'Silhouette Score for DBSCAN: {silhouette_dbscan:.2f}')

# Visualize DBSCAN clustering results
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=dbscan_labels, cmap='viridis')
plt.title('DBSCAN Clustering')
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import linkage, fcluster
from sklearn.datasets import make_blobs

# Step 1: Generate synthetic data
n_samples = 300
n_features = 5  # Number of features
n_clusters = 3  # True number of clusters

# Generate synthetic data with blobs (representing clusters)
data, _ = make_blobs(n_samples=n_samples, n_features=n_features, centers=n_clusters, random_state=42)
df = pd.DataFrame(data, columns=[f'Feature_{i}' for i in range(n_features)])

# Step 2: Data Preprocessing
# Standardize the data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# Step 3: Apply PCA for 2D visualization
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)

# Step 4: K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(df_scaled)

# K-Means Silhouette Score
silhouette_kmeans = silhouette_score(df_scaled, kmeans_labels)

# Detailed results for K-Means
print("K-Means Clustering Results:")
print(f"- Number of clusters: {kmeans.n_clusters}")
print(f"- Inertia (Within-cluster sum of squares): {kmeans.inertia_}")
print(f"- Silhouette Score: {silhouette_kmeans:.2f}")

# Step 5: Hierarchical Clustering
Z = linkage(df_scaled, method='ward')
hierarchical_labels = fcluster(Z, t=3, criterion='maxclust')

# Hierarchical Silhouette Score
silhouette_hierarchical = silhouette_score(df_scaled, hierarchical_labels)

# Detailed results for Hierarchical Clustering
print("\nHierarchical Clustering Results:")
print(f"- Number of clusters formed: {len(set(hierarchical_labels))}")
print(f"- Silhouette Score: {silhouette_hierarchical:.2f}")

# Step 6: DBSCAN Clustering
dbscan = DBSCAN(eps=0.8, min_samples=5)
dbscan_labels = dbscan.fit_predict(df_scaled)

# Handle DBSCAN when there are noise points (label -1 for noise)
num_noise_points = np.sum(dbscan_labels == -1)

# DBSCAN Silhouette Score (only if there are multiple clusters formed)
if len(set(dbscan_labels)) > 1:
    silhouette_dbscan = silhouette_score(df_scaled, dbscan_labels)
    print(f"\nDBSCAN Clustering Results:")
    print(f"- Number of clusters formed: {len(set(dbscan_labels)) - 1} (Excluding noise)")
    print(f"- Number of noise points: {num_noise_points}")
    print(f"- Silhouette Score: {silhouette_dbscan:.2f}")
else:
    silhouette_dbscan = "N/A (Single cluster or all noise)"
    print(f"\nDBSCAN Clustering Results:")
    print(f"- Number of clusters formed: 1 (All points considered as a single cluster or noise)")
    print(f"- Number of noise points: {num_noise_points}")
    print(f"- Silhouette Score: {silhouette_dbscan}")

# Additional insights
print("\n--- Clustering Analysis Insights ---")
print("1. **K-Means:**")
print("   - K-Means performed well in detecting spherical clusters. It works well when clusters are of similar size and well-separated.")
print(f"   - The silhouette score of {silhouette_kmeans:.2f} suggests good clustering performance.")
print("\n2. **Hierarchical Clustering:**")
print("   - Hierarchical clustering produced a good separation with a silhouette score of {silhouette_hierarchical:.2f}.")
print("   - This method works well for capturing nested clusters, especially when hierarchical relationships are important.")
print(f"   - Number of clusters formed: {len(set(hierarchical_labels))}.")
print("\n3. **DBSCAN:**")
print("   - DBSCAN is ideal for identifying clusters of arbitrary shape and handling noise. It performed well for identifying clusters with density-based separation.")
print(f"   - The silhouette score of {silhouette_dbscan} indicates the quality of the DBSCAN clustering. ")
print(f"   - {num_noise_points} noise points were identified.")
print("\n--- Final Thoughts ---")
print("1. K-Means requires a predefined number of clusters, while DBSCAN can discover clusters of varying shapes without specifying the number of clusters.")
print("2. Hierarchical clustering can be more computationally expensive but is useful when dealing with nested cluster structures.")
print("3. Silhouette scores close to 1 for K-Means and Hierarchical clustering indicate well-defined clusters. DBSCAN may have issues when noise points are present.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import linkage, fcluster
from sklearn.datasets import make_blobs

# Step 1: Generate synthetic data
n_samples = 300
n_features = 5  # Number of features
n_clusters = 3  # True number of clusters

# Generate synthetic data with blobs (representing clusters)
data, _ = make_blobs(n_samples=n_samples, n_features=n_features, centers=n_clusters, random_state=42)
df = pd.DataFrame(data, columns=[f'Feature_{i}' for i in range(n_features)])

# Step 2: Data Preprocessing
# Standardize the data
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# Step 3: Apply PCA for 2D visualization
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)

# Step 4: K-Means Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(df_scaled)

# K-Means Silhouette Score
silhouette_kmeans = silhouette_score(df_scaled, kmeans_labels)

# Step 5: Hierarchical Clustering
Z = linkage(df_scaled, method='ward')
hierarchical_labels = fcluster(Z, t=3, criterion='maxclust')

# Hierarchical Silhouette Score
silhouette_hierarchical = silhouette_score(df_scaled, hierarchical_labels)

# Step 6: DBSCAN Clustering
dbscan = DBSCAN(eps=0.8, min_samples=5)
dbscan_labels = dbscan.fit_predict(df_scaled)

# Handle DBSCAN when there are noise points (label -1 for noise)
num_noise_points = np.sum(dbscan_labels == -1)

# DBSCAN Silhouette Score (only if there are multiple clusters formed)
if len(set(dbscan_labels)) > 1:
    silhouette_dbscan = silhouette_score(df_scaled, dbscan_labels)
else:
    silhouette_dbscan = "N/A (Single cluster or all noise)"

# Visualization: Box Plot, Histogram, Scatter Plot
# Box Plot - to visualize the distribution of each feature
plt.figure(figsize=(10, 6))
sns.boxplot(data=df)
plt.title("Box Plot of Features")
plt.show()

# Histogram - to visualize the distribution of data for each feature
plt.figure(figsize=(10, 6))
df.hist(bins=20, figsize=(12, 8), layout=(2, 3))  # Adjust layout based on the number of features
plt.suptitle("Histogram of Features")
plt.tight_layout()
plt.show()

# Scatter Plot - for PCA reduced dimensions
plt.figure(figsize=(10, 6))
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=kmeans_labels, cmap='viridis', label='K-Means', alpha=0.6)
plt.title("K-Means Clustering Results (PCA projection)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar()
plt.show()

# Scatter Plot for Hierarchical clustering
plt.figure(figsize=(10, 6))
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=hierarchical_labels, cmap='plasma', label='Hierarchical', alpha=0.6)
plt.title("Hierarchical Clustering Results (PCA projection)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar()
plt.show()

# Scatter Plot for DBSCAN clustering
plt.figure(figsize=(10, 6))
plt.scatter(df_pca[:, 0], df_pca[:, 1], c=dbscan_labels, cmap='coolwarm', label='DBSCAN', alpha=0.6)
plt.title("DBSCAN Clustering Results (PCA projection)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar()
plt.show()

# Detailed Results and Analysis
print(f"K-Means Silhouette Score: {silhouette_kmeans:.2f}")
print(f"Hierarchical Clustering Silhouette Score: {silhouette_hierarchical:.2f}")
if silhouette_dbscan != "N/A (Single cluster or all noise)":
    print(f"DBSCAN Silhouette Score: {silhouette_dbscan:.2f}")
else:
    print(f"DBSCAN Silhouette Score: {silhouette_dbscan}")

print("\n--- Clustering Analysis Insights ---")
print("1. **K-Means:**")
print(f"   - K-Means Silhouette Score: {silhouette_kmeans:.2f}. This suggests a moderate clustering quality.")
print("   - The clusters are well-separated but might struggle with non-spherical shapes.")

print("\n2. **Hierarchical Clustering:**")
print(f"   - Hierarchical Clustering Silhouette Score: {silhouette_hierarchical:.2f}. This indicates a good clustering structure.")
print("   - It can capture nested clusters well, and is useful for dendrogram-based cluster analysis.")

print("\n3. **DBSCAN:**")
print(f"   - DBSCAN Silhouette Score: {silhouette_dbscan} (could be low if many noise points are detected).")
print("   - DBSCAN is good for arbitrary shaped clusters but may perform poorly with high noise levels.")
print(f"   - Noise points: {num_noise_points}.")
```
