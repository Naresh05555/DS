#2.Encoding Techniques

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Example dataset as a dictionary (replace with your actual data)
data_dict = {
    'Workclass': ['Private', 'Self-emp', 'Private', 'Gov', 'Private'],
    'Education': ['Bachelors', 'HS-grad', 'HS-grad', 'Masters', 'Bachelors'],
    'MaritalStatus': ['Married', 'Single', 'Single', 'Married', 'Divorced'],
    'Occupation': ['Tech-support', 'Craft-repair', 'Craft-repair', 'Other-service', 'Exec-managerial'],
    'Income': ['<=50K', '>50K', '<=50K', '>50K', '<=50K']
}

# Convert the dictionary to a DataFrame
data = pd.DataFrame(data_dict)

# Identify categorical columns
categorical_cols = data.select_dtypes(include=['object']).columns

# Separate columns for One-Hot Encoding (<5 categories) and Label Encoding (>=5 categories)
one_hot_cols = [col for col in categorical_cols if data[col].nunique() < 5]
label_encode_cols = [col for col in categorical_cols if data[col].nunique() >= 5]

# Apply One-Hot Encoding
data = pd.get_dummies(data, columns=one_hot_cols, drop_first=False)

# Apply Label Encoding
label_encoder = LabelEncoder()
for col in label_encode_cols:
    data[col] = label_encoder.fit_transform(data[col])

# Print the transformed dataset
print(data)

#What is encoding technique
#Ans:An encoding technique transforms data from one form to another, typically to ensure it's compatible with various systems, efficient for storage or transmission, or secure against unauthorized access.

#Explanation:
#1. Dataset Representation
#Instead of uploading files, the dataset is represented using a dictionary where the keys are column names (e.g., "Workclass", "Education") and the values are lists containing the data for those columns. This dictionary is then converted into a pandas DataFrame for easier manipulation and processing.


#2.Identifying Categorical Columns
#The next step is to identify which columns in the dataset contain categorical data. These are typically columns where the values are strings or objects (e.g., "Private", "Self-emp", "Married"). This is done by checking the data types of each column, and those with an object type are considered categorical.

#3. Categorizing Columns by Number of Unique Values
#The categorical columns are then divided into two groups based on the number of unique categories they contain:
#One-Hot Encoding Columns: These are the categorical variables that have fewer than 5 unique categories. These columns are suited for One-Hot Encoding because:
#One-Hot Encoding creates a new binary column for each unique category.
#This is appropriate for categorical variables that do not have an inherent order or ranking (e.g., "Workclass" with categories like "Private", "Self-emp").

#Label Encoding Columns:
#These are the categorical variables that have 5 or more unique categories. These columns are suited for Label Encoding because:
#Label Encoding assigns a unique integer value to each category, which is more efficient for variables with many categories.
#It is best used for ordinal variables where there is a meaningful order to the categories (e.g., education levels like "High School", "Bachelors", "Masters").

#4.Applying One-Hot Encoding
#One-Hot Encoding is applied to the columns with fewer than 5 unique categories. For each unique category in a column, a new binary column is created. A 1 is placed in the new column for rows that belong to that category, and a 0 is placed for all other rows. This transformation ensures that no assumptions are made about the ordinal relationship between categories.

#5.Applying Label Encoding
#Label Encoding is applied to the columns with 5 or more unique categories. For each unique category in a column, a unique integer is assigned (e.g., "Private" → 0, "Self-emp" → 1, etc.). This encoding is efficient in terms of memory usage, especially when dealing with high-cardinality variables. However, Label Encoding assumes an ordinal relationship, so it should only be used when the categories have a meaningful order.
```

#3.Feature Engineering

```python
import pandas as pd
import numpy as np

# Example dataset as a dictionary (replace with your actual data)
data_dict = {
    'Workclass': ['Private', 'Self-emp', 'Private', 'Gov', 'Private'],
    'Education': ['Bachelors', 'HS-grad', 'HS-grad', 'Masters', 'Bachelors'],
    'MaritalStatus': ['Married', 'Single', 'Single', 'Married', 'Divorced'],
    'Occupation': ['Tech-support', 'Craft-repair', 'Craft-repair', 'Other-service', 'Exec-managerial'],
    'Income': [45000, 50000, 38000, 60000, 45000],  # Example of skewed numerical feature
    'Age': [25, 32, 28, 45, 58]
}

# Convert the dictionary to a DataFrame
data = pd.DataFrame(data_dict)

# Feature 1: Create a new feature 'Education_Level' based on the 'Education' column
data['Education_Level'] = data['Education'].apply(lambda x: 'Higher Education' if x in ['Bachelors', 'Masters'] else 'Lower Education')

# Feature 2: Create a new feature 'Age_Group' based on the 'Age' column
data['Age_Group'] = pd.cut(data['Age'], bins=[0, 30, 50, 100], labels=['Young', 'Middle-Aged', 'Older'])

# Apply log transformation to 'Income' to deal with skewness
data['Income_Log'] = np.log1p(data['Income'])  # log1p is used to avoid taking log of 0

# Output the transformed dataset
print(data)

#What is Feature Engineering?
#Feature engineering is the process of creating new features or modifying existing ones to improve the performance of machine learning models. Features are the variables or attributes that are used as inputs for these models. By transforming raw data into meaningful features, you can help the model learn patterns more effectively.

#1.Education_Level:
#We apply a lambda function to the Education column to classify people into either "Higher Education" or "Lower Education" based on their education level.
#This feature captures whether a person has a higher education level, which could influence their income level.

#2.Age_Group:
#We use pd.cut() to divide the Age column into three groups: "Young", "Middle-Aged", and "Older".
#This feature helps the model understand patterns related to different age groups, which might impact income or occupation choices.

#3.Log Transformation of Income:
#The Income column is often skewed due to outliers, where a few individuals might earn significantly more than the majority. By applying a log transformation (np.log1p()), we reduce the impact of these outliers and make the distribution more Gaussian.'
#log1p() is used instead of log() to handle cases where Income is 0 (though this is unlikely in most datasets, it's a safe approach).
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Example dataset as a dictionary (replace with your actual data)
data_dict = {
    'Workclass': ['Private', 'Self-emp', 'Private', 'Gov', 'Private'],
    'Education': ['Bachelors', 'HS-grad', 'HS-grad', 'Masters', 'Bachelors'],
    'MaritalStatus': ['Married', 'Single', 'Single', 'Married', 'Divorced'],
    'Occupation': ['Tech-support', 'Craft-repair', 'Craft-repair', 'Other-service', 'Exec-managerial'],
    'Income': [45000, 50000, 38000, 60000, 45000],  # Example of skewed numerical feature
    'Age': [25, 32, 28, 45, 58]
}

# Convert the dictionary to a DataFrame
data = pd.DataFrame(data_dict)

# Feature 1: Create a new feature 'Education_Level' based on the 'Education' column
data['Education_Level'] = data['Education'].apply(lambda x: 'Higher Education' if x in ['Bachelors', 'Masters'] else 'Lower Education')

# Feature 2: Create a new feature 'Age_Group' based on the 'Age' column
data['Age_Group'] = pd.cut(data['Age'], bins=[0, 30, 50, 100], labels=['Young', 'Middle-Aged', 'Older'])

# Apply log transformation to 'Income' to deal with skewness
data['Income_Log'] = np.log1p(data['Income'])  # log1p is used to avoid taking log of 0

# Plot the distribution of 'Income' before and after log transformation
plt.figure(figsize=(12, 6))

# Plot original income distribution
plt.subplot(1, 2, 1)
sns.histplot(data['Income'], kde=True, color='blue', bins=10)
plt.title('Original Income Distribution')

# Plot transformed income distribution (after log transformation)
plt.subplot(1, 2, 2)
sns.histplot(data['Income_Log'], kde=True, color='green', bins=10)
plt.title('Income Distribution After Log Transformation')

plt.tight_layout()
plt.show()

# Plot the distribution of new features ('Education_Level' and 'Age_Group')
plt.figure(figsize=(12, 6))

# Plot Education_Level distribution
plt.subplot(1, 2, 1)
sns.countplot(x=data['Education_Level'], palette='Set2')
plt.title('Education Level Distribution')

# Plot Age_Group distribution
plt.subplot(1, 2, 2)
sns.countplot(x=data['Age_Group'], palette='Set1')
plt.title('Age Group Distribution')

plt.tight_layout()
plt.show()

# Output the transformed dataset
print(data)
```

#4. Feature Selection


```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
import seaborn as sns
import matplotlib.pyplot as plt

# Example dataset as a dictionary (replace with your actual data)
data_dict = {
    'Workclass': ['Private', 'Self-emp', 'Private', 'Gov', 'Private'],
    'Education': ['Bachelors', 'HS-grad', 'HS-grad', 'Masters', 'Bachelors'],
    'MaritalStatus': ['Married', 'Single', 'Single', 'Married', 'Divorced'],
    'Occupation': ['Tech-support', 'Craft-repair', 'Craft-repair', 'Other-service', 'Exec-managerial'],
    'Income': [45000, 50000, 38000, 60000, 45000],  # Example of skewed numerical feature
    'Age': [25, 32, 28, 45, 58]
}

# Convert the dictionary to a DataFrame
data = pd.DataFrame(data_dict)

# Apply Isolation Forest for outlier detection
# We will focus on numerical columns for outlier detection
numerical_data = data[['Income', 'Age']]

# Initialize Isolation Forest model
iso_forest = IsolationForest(contamination=0.2)  # Assume 20% contamination (outliers)
outliers = iso_forest.fit_predict(numerical_data)

# -1 indicates outlier, 1 indicates normal data point
data['Outlier'] = outliers

# Remove outliers
cleaned_data = data[data['Outlier'] == 1].drop('Outlier', axis=1)

# Plot the distribution of 'Income' before and after removing outliers
plt.figure(figsize=(12, 6))

# Before removing outliers
plt.subplot(1, 2, 1)
sns.histplot(data['Income'], kde=True, color='blue', bins=10)
plt.title('Income Distribution (Before Outlier Removal)')

# After removing outliers
plt.subplot(1, 2, 2)
sns.histplot(cleaned_data['Income'], kde=True, color='green', bins=10)
plt.title('Income Distribution (After Outlier Removal)')

plt.tight_layout()
plt.show()

# Output the cleaned data
print(cleaned_data)


```
