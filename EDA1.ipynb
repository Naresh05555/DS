#3.Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the theme for seaborn plots
sns.set_theme(style="whitegrid")

# 1. Histogram: Distribution of a column (e.g., LB)
plt.figure(figsize=(8, 6))
sns.histplot(df_cleaned['LB'], bins=30, kde=True, color='blue')
plt.title('Distribution of LB (Baseline Heart Rate)', fontsize=14)
plt.xlabel('LB (Baseline Heart Rate)')
plt.ylabel('Frequency')
plt.show()

# 2. Boxplot: Highlighting potential outliers for LB
plt.figure(figsize=(8, 6))
sns.boxplot(x=df_cleaned['LB'], color='cyan')
plt.title('Boxplot of LB (Baseline Heart Rate)', fontsize=14)
plt.xlabel('LB (Baseline Heart Rate)')
plt.show()

# 3. Scatterplot: Relationship between LB and AC
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df_cleaned['LB'], y=df_cleaned['AC'], color='green')
plt.title('Scatterplot of LB vs AC', fontsize=14)
plt.xlabel('LB (Baseline Heart Rate)')
plt.ylabel('AC (Acceleration)')
plt.show()

#What is Data Visualization
#Data visualization is the graphical representation of information and data. By using visual elements like charts, graphs, and maps, data visualization tools provide an accessible way to see and understand trends, outliers, and patterns in data.

#Explanation:
#1.Histogram:
#Displays the frequency of values for LB (Baseline Heart Rate).
#The kernel density estimation (KDE) curve provides a smooth approximation of the data distribution.

#2.Boxplot:
#Shows the median, quartiles (25th and 75th), and potential outliers in LB.
#Outliers are marked as points beyond the "whiskers."

#3.Scatterplot:
#Explores the relationship between LB (Baseline Heart Rate) and AC (Acceleration).
#Helps identify correlations or patterns between the two variables.
```

#4.Pattern Recognition and Insights:


```python
from sklearn.cluster import KMeans
import numpy as np

# 1. Correlation Heatmap
plt.figure(figsize=(10, 8))
correlation_matrix = df_cleaned.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap", fontsize=14)
plt.show()

# 2. K-Means Clustering
# Selecting two features for clustering (e.g., LB and AC)
kmeans_data = df_cleaned[['LB', 'AC']].dropna()

# Fit K-Means
kmeans = KMeans(n_clusters=3, random_state=0)
clusters = kmeans.fit_predict(kmeans_data)

# Add cluster labels to the data
kmeans_data['Cluster'] = clusters

# Scatterplot with clusters
plt.figure(figsize=(8, 6))
sns.scatterplot(x='LB', y='AC', hue='Cluster', data=kmeans_data, palette='viridis')
plt.title("K-Means Clustering (LB vs AC)", fontsize=14)
plt.xlabel('LB (Baseline Heart Rate)')
plt.ylabel('AC (Acceleration)')
plt.legend(title="Cluster")
plt.show()

# 3. Pairplot
sns.pairplot(df_cleaned[['LB', 'AC', 'FM', 'UC']])
plt.suptitle("Pairplot of Selected Features", y=1.02, fontsize=14)
plt.show()

#Explanation:
#1.Correlation Heatmap:
#Measures the relationship between numerical variables.
#A correlation coefficient (from -1 to 1) shows how strongly two variables are related.

#2.Clustering (K-Means):
#Groups data points based on similarity.
#Helps identify natural groupings in the data.

#3.Pairplot:
#A matrix of scatterplots showing pairwise relationships among variables.
```

```python
# 1. Extract top correlated features
top_correlations = correlation_matrix.unstack().sort_values(ascending=False)
# Filter to exclude self-correlation (1.0) and redundant pairs
top_correlations = top_correlations[(top_correlations < 1.0) & (top_correlations > 0.5)]

print("\nTop Positive Correlations:")
print(top_correlations)

# 2. Count the number of data points in each cluster
cluster_counts = kmeans_data['Cluster'].value_counts()

print("\nCluster Sizes:")
print(cluster_counts)

# 3. Summarize relationships in pairplot
pairplot_description = """
The pairplot provides a detailed view of how selected variables interact.
Look for linear relationships, clusters, or trends that highlight dependencies.
"""

print(pairplot_description)

#Explanation:
#1.Correlation Insights:
#Extracts feature pairs with strong positive correlations (> 0.5).
#Helps identify variables that may influence each other.

#2.Cluster Analysis:
#Summarizes the size of each cluster to understand data distribution.
#Larger clusters might indicate predominant patterns.

#3.Pairplot Summary:
#Offers visual confirmation of observed relationships and interactions.
```
